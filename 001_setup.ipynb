```python
doctors = pd.DataFrame({
    "doctor": ["Dr. A", "Dr. B", "Dr. C", "Dr. D", "Dr. E"],
    "rx_per_year": [2000, 1800, 150, 1500, 2200],
    "new_class_share": [0.05, 0.45, 0.80, 0.20, 0.50],
    "growth": [0.00, 0.15, 0.10, 0.30, 0.20],
    "allows_visits": [True, True, True, True, False],
})

doctors
```

```python
duckdb.sql("""
    SELECT doctor, rx_per_year, new_class_share
    FROM doctors
    WHERE allows_visits = TRUE
      AND rx_per_year >= 1000
    ORDER BY new_class_share DESC
""").df()
```

```python
doctors.to_csv(f"{base}/data/bronze/practice_doctors.csv", index=False)
print("Saved!")

```
